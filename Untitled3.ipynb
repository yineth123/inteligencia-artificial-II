Clasificación y diagnóstico
Clasificación: el modelo aprende a identificar datos en diferentes categorías.
Datos: información utilizada para entrenar y probar el modelo.
MNIST: conjunto de imágenes de números escritos a mano del 0 al 9.
Objetivo: entrenar el modelo y evaluar qué tan bien clasifica.
Exactitud: indica cuántas predicciones fueron correctas, pero no siempre es la mejor medida de rendimiento.
Matriz de confusión: compara lo real vs. lo predicho y muestra positivos, negativos y errores.
Precisión: de los positivos predichos, cuántos fueron realmente positivos.
Exhaustividad: de los positivos reales, cuántos encontró el modelo.
Umbral de decisión: determina cuándo una predicción se considera positiva o negativa.
Curva ROC: analiza el rendimiento según diferentes umbrales y muestra los falsos positivos.
Clasificación multiclase: cuando existen más de dos categorías, como MNIST.
Matriz de confusión normalizada: muestra en porcentajes cómo clasifica el modelo cada clase.
