CRISP-DM: El plano maestro

CRISP-DM es una metodología para desarrollar proyectos de minería de datos. Es un proceso cíclico e interactivo que permite organizar un proyecto desde el problema hasta su aplicación.

1. Negocio

Proceso: identificar el problema → definir objetivos → establecer metas → crear un plan.
Ejemplo: reducir la fuga de clientes.

2. Datos

Proceso: recopilar datos → conocer su origen → analizar qué información tenemos → comprobar si sirve para el objetivo.

3. Preparación de datos

Proceso: limpiar → corregir errores → eliminar datos innecesarios → organizar → transformar.
Es una de las etapas que requiere más trabajo.

4. Modelado y evaluación

Proceso: elegir técnicas → construir modelos → probar → comparar → evaluar resultados.
Aquí se aplican algoritmos y matemáticas para encontrar patrones y obtener resultados.

5. Despliegue y futuro

Proceso: implementar → integrar → monitorear → generar informes → revisar → mejorar.
Los resultados se llevan a la práctica y pueden evolucionar con tecnologías como la IA.

En resumen:
Negocio → Datos → Preparación → Modelado y evaluación → Despliegue y futuro.
